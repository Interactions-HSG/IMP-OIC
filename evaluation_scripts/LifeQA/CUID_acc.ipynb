```python
def count_total_identifications(video_script):
    total_identifications = set()
    for line in video_script.splitlines():
        words = line.split()
        for word in words:
            if "_" in word:
                total_identifications.add(word)
    print(total_identifications)
    return len(total_identifications)

# Example usage:
video_script = video_description
total_identifications = count_total_identifications(video_script)
print("Total Identifications:", total_identifications)
```

```python
# wrong ids contain duplicates as well
def calculate_metrics(correct_identifications, total_ids, wrong_identification):
    adjusted_accuracy = (correct_identifications) / total_ids
    recall = correct_identifications / (correct_identifications + wrong_identification)
    f1_score = 2 * (adjusted_accuracy * recall) / (adjusted_accuracy + recall)
    return adjusted_accuracy, recall, f1_score

correct_identifications = 35
total_ids = total_identifications
wrong_identification = 10

accuracy, recall, f1_score = calculate_metrics(correct_identifications, total_ids, wrong_identification)

print("Adjusted Accuracy:", accuracy*100)
print("Recall:", recall*100)
print("F1-score:", f1_score*100)
```
