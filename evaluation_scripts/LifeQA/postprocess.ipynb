data = pd.read_csv('/.../.../Developer/IMP-OIC/LifeQA/for_eval/OIC_gpt4_w_st.csv')

```python
import random

q_ids = data['q_id']

for qid in q_ids:
    options = [0, 1, 2, 3]
    que = data.query("q_id=={}".format(qid))
    match = que['Match'].values[0]
    answer_id = que["ans_idx"].values[0]
    OIC_answer = que["OIC_answer"].values[0]
    
    
    if pd.isna(OIC_answer):
        if match == 'Correct':
            data.loc[data['q_id'] == qid, 'OIC_answer'] = answer_id
        else:
            if match == 'Wrong':
                options.remove(answer_id) 
                random_number = random.choice(options)
                data.loc[data['q_id'] == qid, 'OIC_answer'] = random_number
            else:
                data.loc[data['q_id'] == qid, 'OIC_answer'] = match
    else:
        if len(OIC_answer)>1:
            
            data.loc[data['q_id'] == qid, 'OIC_answer'] = OIC_answer
        else:
            if OIC_answer != answer_id:
                data.loc[data['q_id'] == qid, 'Match'] = 'Wrong'
            else:
                data.loc[data['q_id'] == qid, 'Match'] = 'Correct'
            

data.head(20)
```

```python
no_match = data.loc[(data['Match'] != 'Correct') & (data['Match'] != 'Wrong')]
```

```python
list_to_modify = no_match['q_id'].tolist()
options = [0, 1, 2, 3]
for i in list_to_modify:
    random_number = random.choice(options)
    data.loc[data['q_id'] == i, 'OIC_answer'] =  random_number
```

```python
q_ids = data['q_id']

for qid in q_ids:
    que = data.query("q_id=={}".format(qid))
    match = que['Match'].values[0]
    answer_id = que["ans_idx"].values[0]
    OIC_answer = que["OIC_answer"].values[0]
    
    if len(str(OIC_answer))>1:
            data.loc[data['q_id'] == qid, 'Match'] = OIC_answer
    else:
        if int(OIC_answer) != int(answer_id):
            data.loc[data['q_id'] == qid, 'Match'] = 'Wrong'
        else:
            data.loc[data['q_id'] == qid, 'Match'] = 'Correct'
    
data.info
```
