```python

import pandas as pd
import numpy as np
import string
import os
import sys
import shutil

baseline_gpt4_wo_st_ds = pd.read_csv('for_eval/baseline_gpt4_wo_st.csv').loc[:99]
baseline_gpt4_wo_st_ds = baseline_gpt4_wo_st_ds.drop_duplicates()
baseline_gpt4_wo_st_ds.reindex()
baseline_gpt4_wo_st_ds = baseline_gpt4_wo_st_ds.dropna()
baseline_gpt4_wo_st_ds = baseline_gpt4_wo_st_ds.loc[:, ~baseline_gpt4_wo_st_ds.columns.str.contains('^Unnamed')]
baseline_gpt4_wo_st_ds.name = 'baseline without subtitles'

baseline_gpt4_w_st_ds = pd.read_csv('for_eval/baseline_gpt4_w_st.csv').loc[:99]
baseline_gpt4_w_st_ds['subtitle']= baseline_gpt4_w_st_ds['subtitle'].fillna('No subtitle')
baseline_gpt4_w_st_ds = baseline_gpt4_w_st_ds.drop_duplicates()
baseline_gpt4_w_st_ds.reindex()
baseline_gpt4_w_st_ds = baseline_gpt4_w_st_ds.dropna()
baseline_gpt4_w_st_ds = baseline_gpt4_w_st_ds.loc[:, ~baseline_gpt4_w_st_ds.columns.str.contains('^Unnamed')]
baseline_gpt4_w_st_ds.name = 'baseline with subtitles'

OIC_response_gpt4_w_st_ds = pd.read_csv('for_eval/OIC_gpt4_w_st_3.csv').loc[:99]
OIC_response_gpt4_w_st_ds['subtitle']= OIC_response_gpt4_w_st_ds['subtitle'].fillna('No subtitle')
OIC_response_gpt4_w_st_ds = OIC_response_gpt4_w_st_ds.drop_duplicates()
OIC_response_gpt4_w_st_ds.reindex()
OIC_response_gpt4_w_st_ds = OIC_response_gpt4_w_st_ds.dropna()
OIC_response_gpt4_w_st_ds = OIC_response_gpt4_w_st_ds.loc[:, ~OIC_response_gpt4_w_st_ds.columns.str.contains('^Unnamed')]
OIC_response_gpt4_w_st_ds.name = 'OIC response with subtitles'

OIC_response_gpt4_wo_st_ds = pd.read_csv('for_eval/OIC_gpt4_wo_st_3.csv').loc[:99]
OIC_response_gpt4_wo_st_ds['subtitle']= OIC_response_gpt4_wo_st_ds['subtitle'].fillna('No subtitle')
OIC_response_gpt4_wo_st_ds['OIC_context'] = OIC_response_gpt4_wo_st_ds['OIC_context'].replace(OIC_response_gpt4_w_st_ds.set_index('qid')['OIC_context'])
OIC_response_gpt4_wo_st_ds = OIC_response_gpt4_wo_st_ds.drop_duplicates()
OIC_response_gpt4_wo_st_ds.reindex()
OIC_response_gpt4_wo_st_ds = OIC_response_gpt4_wo_st_ds.dropna()
OIC_response_gpt4_wo_st_ds = OIC_response_gpt4_wo_st_ds.loc[:, ~OIC_response_gpt4_wo_st_ds.columns.str.contains('^Unnamed')]
OIC_response_gpt4_wo_st_ds.name = 'OIC response without subtitles'
```

```python
for i in [baseline_gpt4_wo_st_ds, baseline_gpt4_w_st_ds, OIC_response_gpt4_wo_st_ds, OIC_response_gpt4_w_st_ds]:
  print('-'*100)
  count = (i['Match'].value_counts()['Correct'])
  length = len(i)
  print('length of {} is: {}'.format(i.name, length))
  print('Correct: {}'.format(count))
  print('Correct match percentage for {} is: {}'.format(i.name, (count/length)*100))
  
  #ds.to_csv('all_cured.csv')
```

Merge the dataframes for analysis
1. OIC QA compared with Baseline (GPT4) with subtitle

```python
with_st_merged_df = pd.merge(baseline_gpt4_w_st_ds, OIC_response_gpt4_w_st_ds, on= 'qid', suffixes=('_baseline', '_OIC'), how='left')
with_st_merged_df
```

```python
not_matching_answers_w_st = with_st_merged_df.loc[with_st_merged_df.Match_baseline!=with_st_merged_df.Match_OIC, ['qid','OIC_question_OIC','OIC_context_OIC','subtitle_OIC','answer_idx_OIC','OIC_answer_OIC','Match_OIC','OIC_answer_baseline','Match_baseline']]
not_matching_answers_w_st
```

```python
answers_correct_in_OIC_w_st_but_not_in_Baseline = not_matching_answers_w_st.loc[not_matching_answers_w_st.Match_OIC == "Correct"]
answers_correct_in_OIC_w_st_but_not_in_Baseline
```

```python
answers_correct_in_Baseline_w_st_but_not_in_GPT = not_matching_answers_w_st.loc[not_matching_answers_w_st.Match_baseline == "Correct"]
answers_correct_in_Baseline_w_st_but_not_in_GPT
```

```python
answers_same_in_both_baseline_and_OIC_with_st = with_st_merged_df.loc[with_st_merged_df.Match_baseline == with_st_merged_df.Match_OIC]
answers_same_in_both_baseline_and_OIC_with_st
```

```python
answers_wrong_in_both_baseline_and_OIC_w_st = answers_same_in_both_baseline_and_OIC_with_st.loc[answers_same_in_both_baseline_and_OIC_with_st.Match_OIC == 'Wrong']
answers_wrong_in_both_baseline_and_OIC_w_st
```
