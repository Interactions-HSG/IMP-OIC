```python
import pandas as pd

#main_ds_1 = pd.read_csv('new_all_cured_2.csv')
main_ds = pd.read_csv('OIC_accuracy_results.csv')

main_ds =main_ds[main_ds['OIC_answer'] != 0]
#main_ds['OIC_answer_wo_CUID_temp_match'] = main_ds_1['OIC_answer_without_CUID_temp']
main_ds.name = "all"

main_ds.info

#main_ds.to_csv('OIC_accuracy_results.csv')
```

```python

interaction_ds = main_ds.loc[main_ds['question_id'].str.contains('Interaction')]
interaction_ds.name = 'Interaction'
sequence_ds = main_ds.loc[main_ds['question_id'].str.contains('Sequence')]
sequence_ds.name = 'Sequence'
prediction_ds = main_ds.loc[main_ds['question_id'].str.contains('Prediction')]
prediction_ds.name = 'Prediction'
feasibility_ds = main_ds.loc[main_ds['question_id'].str.contains('Feasibility')]
feasibility_ds.name = 'Feasibility'

print('-'*100);
print('OIC Accuracy')
print('-'*100);
for i in [main_ds, interaction_ds, sequence_ds, prediction_ds, feasibility_ds]:
  count = (i['Match'].value_counts()['Correct'])
  length = len(i)
  print('length of {} is: {}'.format(i.name, length))
  print('Correct: {}'.format(count))
  print('Correct match percentage for {} is: {}'.format(i.name, (count / length) * 100))

print('-'*100);
print('OIC without CUID Accuracy')
print('-'*100);
for i in [main_ds, interaction_ds, sequence_ds, prediction_ds, feasibility_ds]:
  count = (i['Wo_CUID_Match'].value_counts()['Correct'])
  length = len(i)
  print('length of {} is: {}'.format(i.name, length))
  print('Correct: {}'.format(count))
  print('Correct match percentage for {} is: {}'.format(i.name, (count / length) * 100))

print('-'*100);
print('RelTR: OIC without CUID and TG Accuracy')
print('-'*100);
for i in [main_ds, interaction_ds, sequence_ds, prediction_ds, feasibility_ds]:
  count = (i['OIC_answer_wo_CUID_temp_match'].value_counts()['Correct'])
  length = len(i)
  print('length of {} is: {}'.format(i.name, length))
  print('Correct: {}'.format(count))
  print('Correct match percentage for {} is: {}'.format(i.name, (count / length) * 100))
```
